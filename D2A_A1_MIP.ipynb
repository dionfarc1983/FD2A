### Full-time Scheduling

```python
# ! DO NOT MODIFY THE FOLLOWING LINES
# * Convert optimized full-time staffing to a 3D array
full_time_staffing = np.zeros((len(employees), len(days), len(shifts)), dtype=int)
for i in range(len(employees)):
    for j in range(len(days)):
        for k in range(len(shifts)):
            full_time_staffing[i, j, k] = int(x[employees[i], days[j], shifts[k]].X)


assert np.all((full_time_staffing.sum(axis=2) > 0)
              <= availability_df.values), 'availability constraint not satisfied'
assert np.all(full_time_staffing.sum(axis=(1, 2)) == 10), 'weekly 40-hours constraint not satisfied'
assert np.all(full_time_staffing.sum(axis=2) <= 2), 'daily 8-hours constraint not satisfied'
assert np.all(full_time_staffing.sum(axis=0) >= 1), 'full-time supervision constraint not satisfied'
```

```python
# ! DO NOT MODIFY THE FOLLOWING LINES
# * Export the schedule in a 2D spreadsheet
full_time_staffing_df = pd.DataFrame(full_time_staffing.reshape(len(employees), -1).T,
                                     index=[f'{d} {s}' for d in days for s in shifts],
                                     columns=employees)
full_time_staffing_df.to_csv('full_time_staffing.csv', index=False, header=False)
full_time_staffing_df
```
